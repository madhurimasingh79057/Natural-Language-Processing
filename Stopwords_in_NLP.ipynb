## **Stopwords using SpaCy**

**Stopwords** are common words that often carry little useful information for certain NLP tasks. Examples include:

the, is, am, are, a, an, and, or, in, on, of, to, for

```python
# Stopwords using spaCy
!pip install spacy
!pip install -m spacy download en_core_web_sm
```

```python
import spacy

nlp = spacy.load("en_core_web_sm")
```

```python
# 1. Check whether a word is a stopword

words = ["the", "is", "Python", "and", "beautiful"]

for word in words:
  doc = nlp(word)

  for token in doc:
    print(token.text, "->", token.is_stop)
```

```python
# 2. Identify stopwords in a sentence

text = "Python is a very useful programming language."
doc = nlp(text)

for token in doc:
  print(token.text, "->", token.is_stop)
```

```python
# 3. Extract only stopwords

text = "This is a simple example of stopwords removal."
doc = nlp(text)

stopwords = []

for token in doc:
  if token.is_stop:
    stopwords.append(token.text)

print(stopwords)
```

```python
# 4. Extract Non-Stopwords

text = "Python is an easy and powerful language."
doc = nlp(text)

words = []

for token in doc:
  if not token.is_stop and not token.is_punct:
    words.append(token.text)

print(words)
```

```python
# 5. Remove stopwords form a sentence

text = "The students are learning data science in the classroom."
doc = nlp(text)

result = []

for token in doc:
  if not token.is_stop:
    result.append(token.text)

print(" ".join(result))
```

```python
# 6. Removing stopwords and punctuation

text = "The students are learning data science in the classroom."
doc = nlp(text)

result = []

for token in doc:
  if not token.is_stop and not token.is_punct:
    result.append(token.text)

print(" ".join(result))
```

```python
# 1. Display NLTK Stopwords

stop_words = stopwords.words("english")
print(stop_words)
```

```python
len(stop_words)
```

```python
# 2. Check whether a word is a stopword

words = ["the", "is", "Python", "and", "beautiful"]

for word in words:
  print(word, "->", word.lower() in stop_words)
```

```python
# 3. Extract stopwords from a sentence

import nltk
nltk.download("stopwords")

from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize

nltk.download("punkt")
nltk.download("punkt_tab")

stop_words = set(stopwords.words("english"))

text = "Python is a very useful programming language."

tokens = word_tokenize(text)

result = []

for word in tokens:
  if word.lower() in stop_words:
    result.append(word)

print(result)
```

```python
# 4. Remove stopwords from a sentence

text = "The students are learning data science."

tokens = word_tokenize(text)

result = []

for word in tokens:
  if word.lower() not in stop_words:
    result.append(word)

print(result)
```

```python
# 5. Remove stopwords and punctuation
import string
# all previous libraries

text = "The students are learning data science!"
tokens = word_tokenize(text)
result = []

for word in tokens:
  if word.lower() not in stop_words and word not in string.punctuation:
    result.append(word)

print(result)
```

```python
# 6. Convert text to lowercase and remove stopwords

text = "Python Is a Powerful Programming Language."
tokens = word_tokenize(text)

result = []

for word in tokens:
  if word.lower() not in stop_words:
    result.append(word.lower())

print(result)
```

```python
# 7. Count stopwords in a Text

text = "This is a simple example to demonstrate stopword removal."
tokens = word_tokenize(text)
count = 0

for word in tokens:
  if word.lower() in stop_words:
    count += 1

print("Number of stopwords:", count)
```

```python
# 9. Add custom stopwords in NLTK

custom_words = {"python","data","language"}
stop_words.update(custom_words)

text = "Python is a data science language."
tokens = word_tokenize(text)
result = []

for word in tokens:
  if word.lower() not in stop_words:
    result.append(word)

print(result)
```
